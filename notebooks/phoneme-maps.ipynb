# Phoneme Maps - UTE-0/0

This notebook explores the phoneme-to-number mappings used in UTE-0/0.

```python
from ute00.phonetics import PhoneticAnalyzer
from ute00.numerology import NumerologyCalculator
from ute00.numerology.systems import NumerologyType

analyzer = PhoneticAnalyzer()
numerology = NumerologyCalculator()
```

## Universal Phoneme Mappings

```python
print("Universal Phoneme Mappings:")
print("="*50)

vowels = "aeiou"
consonants = "bcdfghjklmnpqrstvwxyz"

print("\nVowels:")
for v in vowels:
    m = analyzer.get_phoneme_mapping(v)
    if m:
        print(f"  {v}: real={m.real}, imaginary={m.imaginary}, structural={m.structural}")

print("\nConsonants:")
for c in consonants:
    m = analyzer.get_phoneme_mapping(c)
    if m:
        print(f"  {c}: real={m.real}, imaginary={m.imaginary}, structural={m.structural}")
```

## Numerology Systems Comparison

```python
words = ["water", "fire", "earth", "air", "mother", "father"]

print("Numerology Comparison:")
print("="*70)
print(f"{'Word':<10} {'Pythagorean':>12} {'Kabbalistic':>12} {'Astrological':>12} {'Mirrored':>12}")
print("-"*70)

for word in words:
    comparison = numerology.compare_systems(word)
    print(f"{word:<10} {str(comparison['pythagorean']):>12} {str(comparison['kabbalistic']):>12} {str(comparison['astrological']):>12} {str(comparison['mirrored']):>12}")
```
