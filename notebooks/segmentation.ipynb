# Phoneme Segmentation - UTE-0/0

This notebook demonstrates the phonetic segmentation and analysis capabilities.

```python
from ute00.phonetics import PhoneticAnalyzer, PhoneticSegmenter

analyzer = PhoneticAnalyzer()
segmenter = PhoneticSegmenter()
```

## Phoneme Mapping: Φ(f) = (n_r, n_i, n_s)

Each phoneme maps to a triple:
- n_r = real component (phonetic frequency)
- n_i = imaginary component (symbolism)
- n_s = structural component (position in universal tree)

```python
# Analyze phonemes
text = "universal"
mappings = analyzer.analyze(text)

print(f"Phoneme analysis for '{text}':")
for m in mappings:
    phi = (m.real, m.imaginary, m.structural)
    print(f"  {m.phoneme}: Φ = {phi}")

print(f"\nTotal vibration: {analyzer.get_total_vibration(text)}")
```

## Da Vinci Segmentation

```python
text = "translation"
result = segmenter.segment(text)

print(f"Original: {result.original}")
print(f"Horizontal Inversion: {result.horizontal_inversion}")
print(f"Vertical Inversion: {result.vertical_inversion}")
print(f"Syllables: {result.syllables}")
print(f"Nuclear Vowels: {result.nuclear_vowels}")
```
