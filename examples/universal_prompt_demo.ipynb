# Universal Prompt Demo - 10-Stage Pipeline

This notebook demonstrates the complete 10-stage Universal Translation pipeline.

```python
from universal_translation.core.pipeline import full_pipeline
import json
```

## Run Full Pipeline

```python
# Analyze a sacred text
text = "Om mani padme hum"
result = full_pipeline(text, include_inversions=True)

print(f"Input: {text}")
print("=" * 50)
```

```python
# Stage 1: Segmentation
print("\nStage 1 - Segmentation:")
print(f"  Tokens: {result['stage_1_segmentation']['tokens']}")
print(f"  Syllables: {result['stage_1_segmentation']['syllables']}")
if 'inverted' in result['stage_1_segmentation']:
    print(f"  Inverted: {result['stage_1_segmentation']['inverted']['horizontal']}")
```

```python
# Stage 2: Phonemes
print("\nStage 2 - Phonemes:")
for p in result['stage_2_phonemes']:
    print(f"  {p['token']} -> {p['ipa']} ({p['family']})")
```

```python
# Stage 4: Language Families
print("\nStage 4 - Language Families:")
print(f"  Detected: {result['stage_4_families']}")
```

```python
# Stage 5: Symbolic
print("\nStage 5 - Symbolic:")
for s in result['stage_5_symbolic']:
    print(f"  {s['token']}: {s['symbolic_tags']}")
```

```python
# Stage 6: Archaeo-cosmological
print("\nStage 6 - Archaeo-cosmological:")
print(f"  Synthesis: {result['stage_6_archaeo']['symbolic_synthesis']}")
```

```python
# Stage 7: Numerology
print("\nStage 7 - Numerology:")
print(f"  Total value: {result['stage_7_numerology']['total_value']}")
print(f"  Root: {result['stage_7_numerology']['total_root']}")
```

```python
# Stage 8: Translations
print("\nStage 8 - Multiple Translations:")
for k, v in result['stage_8_translations'].items():
    print(f"  {k}: {v}")
```

```python
# Stage 9: Diagram
print("\nStage 9 - Diagram Data:")
print(json.dumps(result['stage_9_diagram'], indent=2))
```

```python
# Stage 10: Universal Phrase
print("\n" + "=" * 50)
print("Stage 10 - Universal Phrase:")
print(f"\n  {result['stage_10_universal']}")
```
