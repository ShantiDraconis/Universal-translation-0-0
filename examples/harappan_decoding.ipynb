# Harappan Sign Decoding - Demo Notebook

This notebook demonstrates Harappan script decoding using the Universal Translation Engine.

```python
from universal_translation.api.translate import decode_harappan, search_harappan_signs
from universal_translation.core.harappan_engine import (
    decode_sequence,
    encode_sequence,
    get_sign_category,
    get_universal_code,
    decode_to_universal
)
```

## Sign Categories

```python
# Explore sign categories
for sign_id in [1, 8, 41, 75, 109, 417]:
    category = get_sign_category(sign_id)
    code = get_universal_code(sign_id)
    print(f"Sign M-{sign_id:03d}: {category} -> {code}")
```

## Decode a Sequence

```python
# Decode a sequence of signs
sign_ids = [8, 1, 75]  # Fish, Human, Lotus
result = decode_harappan(sign_ids)

print("Decoded Signs:")
for sign in result['decoded_signs']:
    print(f"  {sign['mahadevan_id']}: {sign['category']} -> {sign['universal_code']}")
```

## Full Analysis

```python
# Full analysis of sign sequence
result = decode_harappan([8, 13, 75])
print("Sign text:", result['sign_text'])
print("\nAnalysis:")
print(result['analysis']['synthesis'])
```

## Convert Between Systems

```python
# Encode and decode
symbolic = "SIGN_8 SIGN_13 SIGN_75"
ids = encode_sequence(symbolic)
print(f"Symbolic: {symbolic}")
print(f"IDs: {ids}")
print(f"Back to symbolic: {decode_sequence(ids)}")
```
